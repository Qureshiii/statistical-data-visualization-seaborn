Importing Libraries

```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

plt.style.use('seaborn-v0_8-whitegrid')
```

Importing DataSets

```python
tips = sns.load_dataset('tips')
titanic = sns.load_dataset('titanic')
gap = px.data.gapminder()
iris = px.data.iris()
```

            FacetGrid

```python
sns.catplot(data=tips,kind='violin',x='sex',y='total_bill',col='day',row='time',hue='sex')
```

```python
g = sns.FacetGrid(data=tips,col='day',row='time')

g.map(sns.barplot,'sex','total_bill')
g.add_legend()
```

            Plotting Pairwise Relationship(PairGrid Vs PairPlot)

```python
iris
```

__PairPlot__

```python
sns.pairplot(iris,hue='species',palette='viridis')
```

__PairGrid__

```python
g = sns.PairGrid(data=iris,hue='species',palette='viridis')
# g.map 

g.map(sns.scatterplot)
```

map_diag and map_offdiag

```python
g = sns.PairGrid(data=iris,hue='species',palette='viridis')
# g.map 

g.map_diag(sns.histplot)
g.map_offdiag(sns.scatterplot)
```

map_diag --> map_upper --> map_lower

```python
g = sns.PairGrid(data=iris,hue='species',palette='viridis')
# g.map 

g.map_diag(sns.histplot)
g.map_upper(sns.boxplot)
g.map_lower(sns.scatterplot)
```

            jointGrid Vs jointPlot

__jointPlot__

```python
sns.jointplot(data=tips,kind='hist',x='total_bill',y='tip',hue='sex',palette='viridis')
```

__jointGrid__

```python
g = sns.JointGrid(data=tips,x='total_bill',y='tip',hue='sex')
g.plot(sns.scatterplot,sns.histplot)
```
