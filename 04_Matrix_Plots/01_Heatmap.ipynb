__Heatmap__ : --> Plot rectengular Data as a color encoded matrix.

Importing Libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

plt.style.use('seaborn-v0_8-whitegrid')
```

Importing DataSets

```python
tips = sns.load_dataset('tips')
titanic = sns.load_dataset('titanic')
gap = px.data.gapminder()
```

```python
# this is a axes level fuction so we can use it like in matplotlib

plt.figure(figsize=(12,5))
temp = gap.pivot_table(index='country',columns='year',values='lifeExp')
sns.heatmap(temp,annot=True)
```

if you want this graph only for eurpean countries

```python
# annot parameter

plt.figure(figsize=(15,15))
temp = gap[gap['continent'] =='Europe']
temp = temp.pivot_table(index='country',columns='year',values='lifeExp')
sns.heatmap(temp,annot=True,linewidths=0.5,cmap='viridis')
```
