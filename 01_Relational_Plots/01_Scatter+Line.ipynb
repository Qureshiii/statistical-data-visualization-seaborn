Importing Libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

plt.style.use('default')
```

Importing DataSets

```python
tips = sns.load_dataset('tips')
gap = px.data.gapminder()
```

            1. Relational Plots

__Scatter Plot__

Axes Level Fuction.

```python
sns.scatterplot(data=tips,x='total_bill',y='tip',hue='sex',style='time',size='size',palette='viridis')
```

Figure Level plot

```python
sns.relplot(data=tips,kind='scatter',x='total_bill',y='tip',hue='sex',style='time',size='size',palette='viridis')
```

__Line Plot__

Axes Level function.

```python
temp = gap[gap['country'].isin(['Cambodia','Rwanda','China'])]
sns.lineplot(data=temp,x='year',y='lifeExp',hue='country',palette='Set1')
```

Figure Level Function

```python
temp = gap[gap['country'].isin(['Pakistan','Rwanda','Germany'])]
sns.relplot(data=temp,x='year',y='lifeExp',kind='line',hue='country',style='continent',size='continent')
```

Facet Plot

```python
# facet plot --> figure level plot --> work with relpot
# it will not work with scatter and lineplot. 

sns.relplot(data=tips,kind='scatter',x='total_bill',y='tip',hue='sex',style='time',size='size',col='sex',row='smoker')
```

```python
# col_wrap

sns.relplot(data=gap,x='lifeExp',y='gdpPercap',kind='scatter',col='year',col_wrap=4)
```
