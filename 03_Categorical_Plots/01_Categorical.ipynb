importing libraries

```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

plt.style.use('seaborn-v0_8-whitegrid')
```

importing DataSets

```python
tips = sns.load_dataset('tips')
iris = px.data.iris()
titanic = sns.load_dataset('titanic')
```

            Categorical Scatter Plots

Must one column is categorical and one is numerical.

__Stripplot__

```python
# Axes level function  
# jitter parameter
sns.stripplot(data=tips,x='day',y='total_bill',hue='sex')
```

```python
sns.catplot(data=tips,kind='strip',x='day',y='total_bill',jitter=False,hue='sex')
```

__swarmplot__

```python
sns.swarmplot(data=tips,x='day',y='total_bill',hue='sex')
```

```python
sns.catplot(data=tips,kind='swarm',x='day',y='total_bill',hue='sex')
```

            Distribution Plots

__Boxplot__

```python
# axes level
sns.boxplot(data=tips,x='day',y='total_bill',hue='sex')
```

```python
# figure level
sns.catplot(data=tips,kind='box',x='day',y='total_bill',hue='sex')
```

```python
sns.boxplot(data=tips,y='total_bill')
```

__Violinplot__ == __(Boxplot + KDEplot)__

```python
# axes level
sns.violinplot(data=tips,x='day',y='total_bill')
```

```python
# figure level
sns.catplot(data=tips,kind='violin',x='day',y='total_bill',hue='sex',split=True)
```

            Estimate plots --> For Central tendency

__Barplot__

```python
sns.barplot(data=tips,x='sex',y='total_bill',hue='smoker')
```

```python
sns.catplot(data=tips,kind='bar',x='sex',y='total_bill',hue='smoker',estimator=np.min,errorbar=None)
```

__Pointplot__

```python
sns.catplot(data=tips,kind='point',x='sex',y='total_bill',hue='smoker')
```

__countplot__

```python
sns.catplot(data=tips,kind='box',x='sex',y='total_bill',col='smoker',hue='sex',row='time')
```
