Importing Libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

plt.style.use('seaborn-v0_8-whitegrid')
```

Importing DataSets

```python
tips = sns.load_dataset('tips')
gap = px.data.gapminder()
titanic = sns.load_dataset('titanic')
```

Figure Level Function Plot --> displot

```python
# Plotting Univariate Histogram
sns.displot(data=tips,x='total_bill',kind='hist',bins=20)
```

Axes Level Function Plot --> histplot

```python


sns.histplot(data=tips,x='total_bill')
```

1. it's also possible to visualize the distribution of categorical variable using the logic of a histogram.
 
2. Discrete bins are automatically set for categorical variables. 

```python
# at this place histogram is acting like a countplot. 
sns.displot(data=tips,x='day',kind='hist')
```

```python
# hue parameter
# Element Parameter
sns.displot(data=tips,x='tip',kind='hist',hue='sex',element='step')
```

Histogram on Titanic

```python
sns.displot(data=titanic,x='age',kind='hist',element='step',hue='survived')
```

Facet using col and row is also possible.

```python
sns.displot(data=titanic,x='age',kind='hist',element='step',hue='survived',col='sex')
```

__Bivariate Histogram__ :--> A bivariate histogram bins the data within rectangles that tiles the plot
and then shows the count of observation within each rectangle with the fill color.

```python
sns.displot(data=tips,kind='hist',x='total_bill',y='tip',hue='sex')
```
