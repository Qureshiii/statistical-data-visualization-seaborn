__KDE Plot(Kernal Density Estimation)__ :-> Rather than using discrete bins, a KDE plot smooth the observations with a guassian kernel, producing a continous density estimate.

Importing Libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

plt.style.use('seaborn-v0_8-whitegrid')
```

Importing Libraries

```python
tips = sns.load_dataset('tips')
gap = px.data.gapminder()
titanic = sns.load_dataset('titanic')
```

Axes Level 

```python
sns.kdeplot(data=tips,x='total_bill')
```

Figure Level

```python
# hue
# Fill
sns.displot(data=tips,x='total_bill',kind='kde',hue='sex',fill=True,aspect=2)
```

__Bivariate KDE__ :--> Bivariate KDE plot smoothes the (x,y) observation with a 2D gaussian.

```python
sns.displot(data=tips,kind='kde',x='total_bill',y='tip',hue='sex')
```
