In the Simplest invocation ___regplot___ and ___lmplot___ both the function draw a scatter plot of two variables x and y, and then fit the regression model y ~ x and plot the resulting regression line and a 95% confidence interval for that regression. 

Impoting libraries

```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

plt.style.use('seaborn-v0_8-whitegrid')
```

Importing DataSets

```python
tips = sns.load_dataset('tips')
titanic = sns.load_dataset('titanic')
gap = px.data.gapminder()
```

```python
tips
```

            regplot --> Axes level

```python
# hue parameter is availale in regplot specificaly but it works on lmplot by the way. 
sns.regplot(data=tips,x='total_bill',y='tip')
```

            lmplot --> figure level

```python
sns.lmplot(data=tips,x='total_bill',y='tip',hue='sex')
```

            residplot 

```python
sns.residplot(data=tips,x='total_bill',y='tip')
```
